```python
from pydantic import BaseModel, Field
from typing import List
from llama_index.llms.openai_like import OpenAILike
from llama_index.core import PromptTemplate

# Definisco la classe Pydantic per l'output strutturato
class KeyboardGuide(BaseModel):
    materials: List[str] = Field(description="Elenco dei materiali necessari per costruire una tastiera meccanica")
    steps: List[str] = Field(description="Passaggi descritti punto per punto per costruirla")
    difficulty: str = Field(description="Livello di difficoltà (es. Facile, Medio, Difficile)")

llm = OpenAILike(
    model="Qwen/Qwen3-32B-AWQ",
    api_base="http://10.1.2.98/v1",
    api_key="null",
    is_chat_model=True,
    is_function_calling_model=True,
    timeout=60.0,
    temperature=0.4,
    context_window=12288,
)

# Definisco il template del prompt (senza bisogno di iniettare le istruzioni manualmente)
prompt_template = PromptTemplate(
    "Ragiona a modo e dimmi come si fa una tastiera meccanica."
)

# Uso llm.structured_predict passando direttamente la classe Pydantic
try:
    output = llm.structured_predict(KeyboardGuide, prompt_template)
    print("Dati estratti con successo!\n")
    print(output.model_dump_json(indent=2))
except Exception as e:
    print(f"Errore durante l'estrazione: {e}")
```
