```python
import os

INPUT_FILE = "../Documents/100%claudio/claude2021-2025.txt"
OUTPUT_DIR = "../Documents/100%claudio/singoli/"

os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
import re

SEPARATOR_PATTERN = re.compile(r'^\s*---\s*$', re.MULTILINE)
timestamp_pattern = re.compile(r'(?=\[\d{2}/\d{2}/\d{4},\s*\d{2}:\d{2}\]\s*—)')

with open(INPUT_FILE, encoding="utf-8") as f:
    content = f.read()

# Divisione su righe che contengono solo ---
raw_blocks = SEPARATOR_PATTERN.split(content)
blocks = [b.strip() for b in raw_blocks if b.strip()]

problems = []
for block in blocks:
    if re.search(r'^\[\d{2}/\d{2}/\d{4}', block, re.MULTILINE):
        sub_blocks = timestamp_pattern.split(block)
        problems.extend([s.strip() for s in sub_blocks if s.strip()])
    else:
        problems.append(block)

print(f"Problemi trovati: {len(problems)}")
```

```python
for i, problem in enumerate(problems, start=1):
    out_path = os.path.join(OUTPUT_DIR, f"claude{i}.md")
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(problem + "\n")
    print(f"Scritto: {out_path}")

print("\nDone.")
```

```python
caratteri_max = 0
nomefile = ""
for filename in os.listdir(OUTPUT_DIR):
    if filename.endswith(".md"):
        print(f"Controllando: {filename}")
        with open(os.path.join(OUTPUT_DIR, filename), encoding="utf-8") as f:
            content = f.read()
            num_caratteri = len(content)
            print(f"Caratteri in {filename}: {num_caratteri}")
            if num_caratteri > caratteri_max:
                caratteri_max = num_caratteri
                nomefile = filename

print(f"\nIl file con più caratteri è {nomefile} con {caratteri_max} caratteri.")
```
